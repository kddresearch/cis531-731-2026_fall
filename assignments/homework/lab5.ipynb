```python
import warnings
warnings.filterwarnings("ignore")
from transformers import pipeline
import torch

print(f"PyTorch Version: {torch.__version__} | CUDA Available: {torch.cuda.is_available()}")
```

### 1. Initialize the Zero-Shot Classification Pipeline
We will load the `facebook/bart-large-mnli` model. This model has been trained on the Multi-Genre Natural Language Inference (MNLI) dataset, allowing it to classify text into arbitrary categories without requiring explicit fine-tuning on those specific labels.

```python
# Initialize the pipeline (this may take a moment to download weights on first run)
print("Loading BART Zero-Shot Classifier...")
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
print("Model loaded successfully.")
```

### 2. Define the Corpus and Trope Labels
We will evaluate three distinct transcript chunks against a custom ontology of literary tropes. Notice that the model has never seen these exact labels during its primary training.

```python
# Simulated ASR Chunks (Including our LibriVox WWI Poem)
corpus = [
    "The protagonist stared at the glowing sword, realizing it was the exact weapon described in the ancient prophecy.",
    "The caller sounded devastated, mentioning that the summer romance ended abruptly when they left for college.",
    "What passing-bells for these who die as cattle? Only the monstrous anger of the guns."
]

# Define our custom Trope Ontology
candidate_labels = ["Chosen One Fantasy", "Doomed Youth / Tragedy", "Summer Romance", "Sci-Fi Cyberpunk"]

print(f"Ontology loaded with {len(candidate_labels)} trope categories.")
```

### 3. Execute Classification
We pass the corpus and the labels to the pipeline. The model will return a probability distribution indicating how strongly it believes each text chunk aligns with each trope.

```python
print("Executing Zero-Shot Classification...\n")

for i, text in enumerate(corpus):
    # Execute the pipeline
    result = classifier(text, candidate_labels)
    
    print(f"--- Transcript Chunk {i+1} ---")
    print(f"Text: '{text}'")
    
    # Print the top 2 predicted tropes and their confidence scores
    print(f"Primary Trope: {result['labels'][0]} (Score: {result['scores'][0]:.4f})")
    print(f"Secondary Trope: {result['labels'][1]} (Score: {result['scores'][1]:.4f})")
    print("----------------------------\n")
```
