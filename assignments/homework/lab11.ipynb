## Part 1: FAIR Data Principles & Provenance Tracking

Before presenting any data, we must ensure its lineage is trackable and interoperable. Use the cell below to document the hash, source URI, and access constraints of your mock dataset.

```python
import hashlib
import json
from datetime import datetime

def log_provenance(dataset_name, source_uri, license_class):
    provenance_record = {
        "dataset": dataset_name,
        "uri": source_uri,
        "license": license_class,
        "timestamp": datetime.now().isoformat(),
        "compliance": "FAIR-aligned"
    }
    print("Provenance Logged:", json.dumps(provenance_record, indent=2))

log_provenance(
    dataset_name="Term_Project_Silver_v1",
    source_uri="s3://kdd-lab/term-projects/2026/silver_data.parquet",
    license_class="instructor_provided"
)
```

## Part 2: Dashboard Visualization & Cognitive Load Reduction

Following Wilke's *Fundamentals of Data Visualization*, we will generate a colorblind-safe, high data-ink ratio plot. Notice the removal of unnecessary chart junk (spines, gridlines) and the use of direct labeling.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate mock model performance data
models = ['Baseline', 'Model A', 'Model B (Proposed)']
f1_scores = [0.65, 0.78, 0.84]
ci_lower = [0.62, 0.75, 0.81]
ci_upper = [0.68, 0.81, 0.87]

fig, ax = plt.subplots(figsize=(8, 5))

# Use Okabe-Ito colorblind-safe palette
colors = ['#E69F00', '#56B4E9', '#009E73']

for i, model in enumerate(models):
    ax.errorbar(f1_scores[i], i, xerr=[[f1_scores[i] - ci_lower[i]], [ci_upper[i] - f1_scores[i]]], 
                fmt='o', color=colors[i], markersize=10, elinewidth=2, capsize=5)
    ax.text(ci_upper[i] + 0.02, i, f'{f1_scores[i]:.2f}', va='center', color=colors[i], fontweight='bold')

ax.set_yticks(range(len(models)))
ax.set_yticklabels(models)
ax.set_xlabel('Macro-F1 Score (95% CI)')
ax.set_title('Model Performance Comparison', loc='left', pad=15, fontweight='bold')

# Reduce cognitive load (remove spines)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)

plt.tight_layout()
plt.show()
```
